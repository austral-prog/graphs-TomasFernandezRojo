# Data Visualization Assignment

In this notebook I made 5 different graphs using matplotlib. I used numpy to generate the data and tried different types of plots.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.makedirs('imgs', exist_ok=True)
```

## Graph 1 - Line Plot

This graph shows the temperature in Buenos Aires over 12 months. I chose a line plot because it works well for data that changes over time. You can clearly see that it's hotter in summer (january/february) and colder in winter (june/july).

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temp = [28, 26, 24, 19, 15, 11, 10, 13, 16, 20, 23, 27]

plt.figure(figsize=(10, 5))
plt.plot(months, temp, color='tomato', marker='o', linewidth=2)
plt.title('Average Temperature in Buenos Aires (2024)')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('imgs/line_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Graph 2 - Bar Chart

This bar chart shows how many hours per week I study each subject. I used a bar chart because it's easy to compare different categories. Programming and Math take the most time.

```python
subjects = ['Math', 'Programming', 'Physics', 'English', 'History']
hours = [9, 11, 8, 4, 3]
colors = ['steelblue', 'seagreen', 'mediumpurple', 'orange', 'salmon']

plt.figure(figsize=(8, 5))
plt.bar(subjects, hours, color=colors, edgecolor='black', width=0.5)
plt.title('Weekly Study Hours per Subject')
plt.xlabel('Subject')
plt.ylabel('Hours per week')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('imgs/bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Graph 3 - Histogram

This histogram shows the distribution of exam scores from 100 students. I used a histogram because it's the best way to see how values are spread out. Most students scored between 60 and 80.

```python
np.random.seed(5)
scores = np.random.normal(loc=65, scale=12, size=100)
scores = np.clip(scores, 0, 100)

plt.figure(figsize=(8, 5))
plt.hist(scores, bins=15, color='steelblue', edgecolor='black')
plt.axvline(scores.mean(), color='red', linestyle='--', label=f'Mean: {scores.mean():.1f}')
plt.title('Distribution of Exam Scores (100 students)')
plt.xlabel('Score (points)')
plt.ylabel('Number of students')
plt.legend()
plt.tight_layout()
plt.savefig('imgs/histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Graph 4 - Scatter Plot

This scatter plot shows the relationship between hours studied and the score on the exam. I chose a scatter plot because it helps visualize if there is a correlation between two variables. In this case you can see that more study hours tend to give better results.

```python
np.random.seed(10)
study_hours = np.random.uniform(1, 15, 50)
exam_score = 40 + 3 * study_hours + np.random.normal(0, 8, 50)
exam_score = np.clip(exam_score, 0, 100)

plt.figure(figsize=(8, 5))
plt.scatter(study_hours, exam_score, color='darkorange', edgecolors='black', s=60)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Hours studied')
plt.ylabel('Exam score (points)')
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig('imgs/scatter_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Graph 5 - Pie Chart

This pie chart shows how students spend their free time on different activities. I used a pie chart because it's useful to show proportions out of a total. You can see that most free time goes to sports and social media.

```python
activities = ['Sports', 'Social media', 'Gaming', 'Reading', 'Other']
time_spent = [30, 25, 20, 15, 10]
colors = ['#4CAF50', '#2196F3', '#FF9800', '#9C27B0', '#607D8B']
explode = (0.05, 0.05, 0.05, 0.05, 0.05)

plt.figure(figsize=(7, 6))
plt.pie(time_spent, labels=activities, colors=colors,
        autopct='%1.1f%%', explode=explode, startangle=90)
plt.title('How Students Spend Their Free Time')
plt.tight_layout()
plt.savefig('imgs/custom_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
